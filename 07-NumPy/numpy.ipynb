```python
arr1 = np.array([1, 2, 3])

arr2 = np.array([4, 5, 6])

arr = np.concatenate((arr1, arr2))

print(arr)
```

```python
arr1 = np.array([[1, 2], [3, 4]])

arr2 = np.array([[5, 6], [7, 8]])

arr = np.concatenate((arr1, arr2), axis=1)

print(arr)
print(arr.shape)
```

```python
harr= np.concatenate((arr1, arr2), axis=0)
print(harr)
print(harr.shape)
```

```python
stac=np.stack((arr1,arr2),axis=1)
print(stac)
print(stac.shape)
```

```python
hstac=np.hstack((arr1,arr2))
print(hstac)
print(hstac.shape)
```

```python
vstac=np.vstack((arr1,arr2))
print(vstac)
print(vstac.shape)
```

```python
dstac=np.dstack((arr1,arr2))
print(dstac)
print(dstac.shape)
```
