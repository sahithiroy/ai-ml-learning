```python
#Let's Move to the functions in NUmpy
''' Copy VS View
    The main difference between a copy and a view of an array is that the copy is a new array, and the view is just a view of the original array.

    The copy owns the data and any changes made to the copy will not affect original array, and any changes made to the original array will not affect the copy.

    The view does not own the data and any changes made to the view will affect the original array, and any changes made to the original array will affect the view.
'''
arr = np.array([1, 2, 3, 4, 5])
x = arr.copy()
arr[0] = 42

print(arr)
print(x)
x[2]=33
print(arr)
print(x)
```

```python
arr = np.array([1, 2, 3, 4, 5])
x = arr.view()
arr[0] = 42

print(arr)
print(x)
x[2]=33
print(arr)
print(x)
```

```python
#Joining Arrays using Concatenate
arr1 = np.array([1, 2, 3])

arr2 = np.array([4, 5, 6])

arr = np.concatenate((arr1, arr2))

print(arr)
```

```python
#Joining Arrays using axis like rowwise and column wise
arr1 = np.array([[1, 2], [3, 4]])

arr2 = np.array([[5, 6], [7, 8]])

arr = np.concatenate((arr1, arr2), axis=0)

print(arr)
print(np.concatenate((arr1, arr2), axis=1))
```

```python
#Joining Arrays Using Stack Functions
#Stacking is same as concatenation, the only difference is that stacking is done along a new axis.
#arr1 = np.array([1, 2, 3])

#arr2 = np.array([4, 5, 6])
arr1 = np.array([[1, 2], [3, 4]])

arr2 = np.array([[5, 6], [7, 8]])
arr = np.stack((arr1, arr2), axis=1)

print(arr)
print(np.stack((arr1, arr2), axis=0))
```

```python
#Stacking along rows
arr1 = np.array([1, 2, 3])

arr2 = np.array([4, 5, 6])

arr = np.hstack((arr1, arr2))

print(arr)
```

```python
#stacking along columns
arr1 = np.array([1, 2, 3])

arr2 = np.array([4, 5, 6])

arr = np.vstack((arr1, arr2))

print(arr)
```

```python
#stacking along depth
arr1 = np.array([1, 2, 3])

arr2 = np.array([4, 5, 6])

arr = np.dstack((arr1, arr2))

print(arr)
```

```python
'''
  what's the difference between loc and iloc
  loc:Selects data based on the labels (names) of rows and columns.
  iloc:Selects data based on the integer positions (indices, starting from 0) of rows and columns.
'''
data = {'col1': [10, 20, 30], 'col2': [40, 50, 60]}
df = pd.DataFrame(data, index=['A', 'B', 'C'])
print(df)
# Using loc
print(df.loc['A', 'col1'])  # Selects value at row 'A', column 'col1'

# Using iloc
print(df.iloc[0, 0])
```

```python
'''Consider a DataFrame with columns like Name, Math, and Science. Using melt(),
 you can transform it so that Name remains an identifier, and Math and Science
  are unpivoted into a new Subject column and a Score column.

  Pivot-->Rows to columns when convert the rows into columns one column into multiple cloumns.we used Present information in a new perspective form
  UnPivot -->Collumns to rows eaiser to anlaysis of the dataframe .we used in the easier to perform on a row-by-row basis
'''

data = {'Name': ['Alice', 'Bob'],
        'Math': [90, 85],
        'Science': [88, 92]}
df = pd.DataFrame(data)
print(df)
# Melt the DataFrame
melted_df = pd.melt(df, id_vars=['Name'], var_name='Subject', value_name='Score')

print('melted df',melted_df)
```

```python
'''Use pandas.pivot() when your data strictly adheres to a one-to-one mapping
between the index and columns values, and you don't need to perform any aggregation.'''
#Need to aggregate data
print(melted_df.pivot(index='Name',columns='Subject',values='Score'))
```
