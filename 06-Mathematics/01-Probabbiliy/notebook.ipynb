COnditional probabilty

At its core, conditional probability is about one simple idea: shrinking your universe.

When you ask for a standard probability P(A), you are asking, "Out of everything that could possibly happen, how likely is A?"

When you ask for a conditional probability P(A∣B)—read as "the probability of A given B"—you receive new evidence: B has already happened. Suddenly, all outcomes where B didn't happen no longer exist. Your "universe" shrinks down to only the space where B is true.

Central Limit Theorem
The Central Limit Theorem (CLT) is one of the most important theorems in statistics and machine learning.In simple terms, the Central Limit Theorem states that if you take enough random samples from any population distribution (no matter how weird, skewed, or non-normal it is), the average (mean) of those samples will always form a smooth, symmetrical Normal Distribution (a Bell Curve).1. The Core Idea: "Averages Behave Better Than Individuals"Imagine a single individual outcome in the real world:An individual customer's wait time might be extremely wild or skewed (most wait 1 minute, but a few wait 45 minutes).Rolling a single die gives a uniform, flat distribution (1, 2, 3, 4, 5, 6 all have equal $1/6$ chance).However, if you take groups of people or roll groups of dice and calculate the average of each group:Sample Average #1: $3.2$Sample Average #2: $3.6$Sample Average #3: $3.5$...repeat 1,000 times.If you plot a histogram of all those sample averages, the resulting shape will form a perfect Bell Curve centered right at the true population mean.2. The Mathematical Rules of CLTLet $X_1, X_2, \dots, X_n$ be a sequence of independent and identically distributed (i.i.d.) random variables with true population mean $\mu$ and variance $\sigma^2$.As the sample size $n$ increases (typically $n \ge 30$), the sample mean $\bar{X}$ converges in distribution to a Normal Distribution:$$\bar{X} \sim \mathcal{N}\left(\mu, \frac{\sigma^2}{n}\right)$$Key Insights from the Formula:Center ($\mu$): The average of your sample averages will equal the true population mean $\mu$.Spread/Uncertainty ($\frac{\sigma}{\sqrt{n}}$): As your sample size $n$ gets larger, the spread gets smaller (the standard error shrinks). Larger samples yield far more precise estimates.3. Why the Central Limit Theorem Matters in Machine LearningWithout the CLT, modern machine learning and data science would struggle to perform reliable statistical inference. Here is why:A. A/B Testing & Model EvaluationWhen testing a new ML model or website design, individual user behavior is wildly skewed. Thanks to CLT, we know the mean conversion rate of sample groups follows a Normal distribution. This allows us to use standard statistical tests ($z$-tests, $t$-tests) to prove if Model A is truly better than Model B.B. Confidence IntervalsCLT allows us to say things like: "We are 95% confident that our model's true accuracy is between 88% and 92%," even if the underlying data distribution is unknown.C. Bootstrapping & Ensemble Learning (Bagging)Algorithms like Random Forests combine predictions from dozens of individual decision trees. The average prediction across these trees relies on the principle of CLT to reduce overall model variance and error.

Maximum Likelihood Estimation (MLE) and Maximum A Posteriori (MAP) are two ways to estimate unknown parameters (like probability or model weights) from observed data.The core difference is that MLE relies only on the observed data, while MAP incorporates prior knowledge before looking at the data.       MLE = Likelihood alone (Data only)
       MAP = Likelihood × Prior (Data + Background knowledge)
1. Core IntuitionMLE asks: "Which parameter value makes the data I observed most likely to occur?"MAP asks: "Which parameter value is most probable, given both the data I observed AND my prior beliefs about the world?"If you have a massive amount of data, MLE and MAP converge to the exact same result because strong empirical evidence outweighs prior assumptions. MAP is most useful when sample sizes are small and MLE risks overfitting.2. Simple Example: The Coin Flip ExperimentImagine you find a coin on the street. You flip it 3 times, and it lands on Heads all 3 times.Let $\theta$ be the probability that the coin lands on Heads:A. The MLE Approach (Data Only)MLE looks strictly at the $3$ flips:$$\hat{\theta}_{\text{MLE}} = \frac{\text{Number of Heads}}{\text{Total Flips}} = \frac{3}{3} = \mathbf{1.0}$$MLE Conclusion: The probability of getting heads is $100\%$. The coin is completely two-headed.The Problem: $3$ flips is a tiny sample size. Declaring a coin $100\%$ biased based on $3$ flips is severe overfitting.B. The MAP Approach (Data + Prior Belief)In real life, you know most coins minted by central banks are roughly fair ($50\%$ heads / $50\%$ tails). This domain knowledge is your Prior.We express this prior using pseudo-observations (e.g., assuming we've already seen $2$ heads and $2$ tails in our life before tossing this coin):$$\hat{\theta}_{\text{MAP}} = \frac{\text{Observed Heads} + \text{Prior Heads}}{\text{Total Flips} + \text{Prior Total}} = \frac{3 + 2}{3 + 4} = \frac{5}{7} \approx \mathbf{0.714}$$MAP Conclusion: The coin seems somewhat biased towards heads ($71.4\%$), but the prior pulls the extreme estimate away from $100\%$ down to a more realistic value.3. Mathematical ComparisonFeatureMaximum Likelihood Estimation (MLE)Maximum A Posteriori (MAP)Formula$\hat{\theta}_{\text{MLE}} = \arg\max_\theta P(X \mid \theta)$$\hat{\theta}_{\text{MAP}} = \arg\max_\theta P(X \mid \theta) P(\theta)$Components UsedLikelihood $P(X \mid \theta)$Likelihood $P(X \mid \theta) \times$ Prior $P(\theta)$Philosophical SchoolFrequentistBayesianRiskOverfits easily on small datasetsCan be biased if the choice of prior is badML Machine EquivalentUnregularized Models (e.g., Standard Linear Regression)Regularized Models (e.g., Ridge / Lasso Regression)4. How MAP Maps to Machine Learning RegularizationWhen training machine learning models (like Neural Networks or Linear Regression), minimizing standard empirical loss corresponds to MLE.When you add Regularization (weight decay), you are performing MAP:$$\text{MAP Objective} = \underbrace{\text{Loss Function (Data Likelihood)}}_{\text{MLE}} + \underbrace{\text{Penalty Term (Prior distribution on weights)}}_{\text{Prior}}$$$L_2$ Regularization (Ridge): Equivalent to MAP with a Gaussian Prior (assumes weights are centered around zero).$L_1$ Regularization (Lasso): Equivalent to MAP with a Laplace Prior (assumes most weights are zero).

```python
'''Bernoulli Distribution: A discrete probability distribution for a random variable that can take only two possible outcomes, often labeled as "success" and "failure". The probability of success is denoted by $p$, and the probability of failure is $1-p$. The Bernoulli distribution is a special case of the binomial distribution with a single trial.'''
def bernoulli_distribution(p, n_samples):
    """
    Generate samples from a Bernoulli distribution.

    Parameters:
    p (float): Probability of success (0 <= p <= 1).
    n_samples (int): Number of samples to generate.

    Returns:
    np.ndarray: Array of samples (0s and 1s).
    """
    if n_samples not in [0, 1]:
        return 0.0
    return p**n_samples*(1-p)**(1-n_samples)
print(f"Bernoulli Distribution (p=0.7, n_samples=1): {bernoulli_distribution(0.7, 1)}")
print(f"Bernoulli Distribution (p=0.7, n_samples=0): {bernoulli_distribution(0.7, 0)}")
```

```python
'''Gaussian (Normal) Distribution: A continuous probability distribution characterized by its bell-shaped curve, defined by its mean ($\mu$) and standard deviation ($\sigma$). The Gaussian distribution is widely used in statistics and machine learning due to the Central Limit Theorem, which states that the sum of many independent random variables tends toward a Gaussian distribution, regardless of the original distributions. '''
def gaussianNormal_distribution(mu, sigma, n_samples):
    """
    Generate samples from a Gaussian (Normal) distribution.

    Parameters:
    mu (float): Mean of the distribution.
    sigma (float): Standard deviation of the distribution.
    n_samples (int): Number of samples to generate.

    Returns:
    np.ndarray: Array of samples from the Gaussian distribution.
    """
    denom=np.sqrt(2 * np.pi * sigma**2)
    exponent=np.exp(-((n_samples - mu)**2) / (2 * sigma**2))
    return (1 / denom) * exponent
print(f"Gaussian Distribution (mu=0, sigma=1, n_samples=0): {gaussianNormal_distribution(0, 1, 0)}")
print(f"Gaussian Distribution (mu=0, sigma=1, n_samples=1): {gaussianNormal_distribution(0, 1, 1)}")
import numpy as np

def gaussian_pdf(x: np.ndarray, mean: float, std: float) -> np.ndarray:
    denom = std * np.sqrt(2 * np.pi)
    num = np.exp(-0.5 * ((x - mean) / std) ** 2)
    return num / denom

# Simulate model inference latency (mean = 45ms, std = 5ms)
latencies = np.array([35.0, 42.0, 45.0, 50.0, 60.0])
pdf_values = gaussian_pdf(latencies, mean=45.0, std=5.0)

for lat, pdf in zip(latencies, pdf_values):
    print(f"Latency: {lat}ms | PDF Density: {pdf:.5f}")
```

```python
'''Expecation & Variance: Expectation (or expected value) is the long-term average value of a random variable, calculated as the weighted sum of all possible values, each multiplied by its probability. Variance measures the spread or dispersion of a random variable's values around its expected value, indicating how much the values deviate from the mean.
'''
import numpy as np
values=np.array([1,2,3,4,5])
probablities=np.array([0.1,0.2,0.3,0.2,0.2])
expected_value=np.sum(values*probablities)
variance=np.sum(probablities*(values-expected_value)**2)
print(f"Expected Value: {expected_value}")
print(f"Variance: {variance}")
```

```python
'''Joint Probability Distribution: A joint probability distribution describes the probability of two or more random variables occurring simultaneously. It provides a way to understand the relationship between multiple random variables and how they interact with each other. The joint distribution can be represented in various forms, such as tables, graphs, or mathematical functions, and is essential for analyzing multivariate data and making predictions based on multiple factors.
Marginal Probability Distribution: The marginal probability distribution of a subset of random variables is obtained by summing or integrating the joint probability distribution over the other variables. It provides the probabilities of individual random variables without considering the others, allowing for a focused analysis of specific variables within a multivariate context.
Conditional Probability Distribution: The conditional probability distribution describes the probability of one random variable given the occurrence of another random variable. It is derived from the joint probability distribution and is crucial for understanding dependencies between variables, enabling predictions and inferences based on observed data.'''
import numpy as np

# Joint Probability Matrix P(X=Device, Y=Error)
# Rows (X): [0: Mobile, 1: Desktop]
# Cols (Y): [0: Success, 1: Error]
joint_P = np.array([
    [0.50, 0.10],  # Mobile
    [0.35, 0.05]   # Desktop
])

# 1. Marginal Probability P(Device = Mobile) -> sum over columns
p_mobile = np.sum(joint_P[0, :])

# 2. Joint Probability P(Mobile AND Error)
p_mobile_and_error = joint_P[0, 1]

# 3. Conditional Probability P(Error | Mobile)
p_error_given_mobile = p_mobile_and_error / p_mobile

print(f"P(Mobile): {p_mobile:.2f}")
print(f"P(Error | Mobile): {p_error_given_mobile:.4f}")
```

```python
'''Bayes' Rule: Bayes' Rule is a fundamental theorem in probability theory that describes how to update the probability of a hypothesis based on new evidence. It relates the conditional and marginal probabilities of random events, allowing for the calculation of posterior probabilities given prior knowledge and observed data. Bayes' Rule is widely used in various fields, including statistics, machine learning, and decision-making processes, to make informed predictions and decisions under uncertainty. The formula for Bayes' Rule is given by:
P(H|D) = P(D|H) * P(H) / P(D)
Where:
- P(H|D) is the posterior probability of the hypothesis given the data
- P(D|H) is the likelihood of the data given the hypothesis
- P(H) is the prior probability of the hypothesis
- P(D) is the marginal probability of the data
Bayesian Updating :Bayesian Updating is the iterative process of taking your computed Posterior probability from step 1 and making it your new Prior for step 2 when new evidence arrives.
[ Initial Prior ] ──► + [ Evidence 1 ] ──► [ Posterior 1 ]
                                                  │
                                                  ▼
 [ New Prior ]     ──► + [ Evidence 2 ] ──► [ Posterior 2 ]
'''
def bayes_rule(prior_h, likelihood_d_given_h, marginal_d):
    """
    Calculate the posterior probability using Bayes' Rule.

    Parameters:
    prior_h (float): Prior probability of hypothesis H.
    likelihood_d_given_h (float): Likelihood of data D given hypothesis H.
    marginal_d (float): Marginal probability of data D.

    Returns:
    float: Posterior probability P(H|D).
    """
    return (likelihood_d_given_h * prior_h) / marginal_d
```

```python
def bayesian_update(prior: float, likelihood: float, p_evidence: float) -> float:
    """Updates prior probability given new evidence."""
    posterior = (likelihood * prior) / p_evidence
    return posterior

# Example: Medical Diagnosis Model
# Prior P(Disease) = 1%
p_disease = 0.01

# Test Accuracy (Likelihood): P(Positive Test | Disease) = 99%
p_pos_given_disease = 0.99

# False Positive Rate: P(Positive Test | No Disease) = 5%
p_pos_given_no_disease = 0.05

# Total Evidence Probability P(Positive Test) using Law of Total Probability
p_positive_test = (p_pos_given_disease * p_disease) + (p_pos_given_no_disease * (1 - p_disease))

# Posterior P(Disease | Positive Test)
p_disease_given_pos = bayesian_update(p_disease, p_pos_given_disease, p_positive_test)

print(f"Prior Probability of Disease:     {p_disease * 100:.1f}%")
print(f"Posterior Probability after Test: {p_disease_given_pos * 100:.2f}%")
```

```python
'''Law of Total Probability: The Law of Total Probability is a fundamental theorem in probability theory that provides a way to compute the probability of an event by considering all possible ways that event can occur, based on a partition of the sample space. It states that if we have a set of mutually exclusive and exhaustive events, the probability of any event can be expressed as the sum of the probabilities of that event occurring under each partition, weighted by the probabilities of the partitions themselves. Mathematically, it can be expressed as:
P(A) = Σ P(A | B_i) * P(B_i)'''
def law_of_total_probability(partitions, conditional_probs):
    """
    Calculate the total probability of an event A using the Law of Total Probability.

    Parameters:
    partitions (list): List of probabilities for each partition B_i.
    conditional_probs (list): List of conditional probabilities P(A | B_i) for each partition.

    Returns:
    float: Total probability P(A).
    """
    total_prob = sum(p * cp for p, cp in zip(partitions, conditional_probs))
    return total_prob
partitions = [0.6, 0.4]  # P(B1), P(B2)
conditional_probs = [0.7, 0.2]  # P(A|B1), P(A|B2)
total_probability = law_of_total_probability(partitions, conditional_probs)
print(f"Total Probability P(A): {total_probability:.4f}")
```

```python
'''MarkovChains: A Markov Chain is a mathematical system that undergoes transitions from one state to another within a finite or countable number of possible states. It is characterized by the Markov property, which states that the future state depends only on the current state and not on the sequence of events that preceded it. Markov Chains are widely used in various fields, including statistics, economics, and machine learning, for modeling random processes and predicting future states based on current information. They can be represented using transition matrices, where each entry indicates the probability of moving from one state to another.
'''
import numpy as np

# Transition Matrix T [2x2] for User State: [0: Inactive, 1: Active]
T = np.array([
    [0.85, 0.15],  # P(Inactive->Inactive)=0.85, P(Inactive->Active)=0.15
    [0.10, 0.90]   # P(Active->Inactive)=0.10,   P(Active->Active)=0.90
])

# Initial state distribution: 100% of users start Inactive
state = np.array([1.0, 0.0])

# Simulate state distribution evolution over 50 steps
for step in range(50):
    state = np.dot(state, T)

print(f"State distribution after 50 steps: Inactive={state[0]:.4f}, Active={state[1]:.4f}")

# Solve for exact Stationary Distribution pi using eigenvectors
# pi * T = pi  =>  T.T * pi = pi
eigenvalues, eigenvectors = np.linalg.eig(T.T)
# Find eigenvector corresponding to eigenvalue 1.0
stationary_vector = eigenvectors[:, np.isclose(eigenvalues, 1.0)].flatten().real
stationary_dist = stationary_vector / np.sum(stationary_vector)

print(f"Analytical Stationary Distribution: Inactive={stationary_dist[0]:.4f}, Active={stationary_dist[1]:.4f}")
```
