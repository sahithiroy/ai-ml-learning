```python
class QuickSort:
    def partition(self,arr,low,high):
        pivot=arr[high]
        i=low-1
        for j in range(low,high):
            if arr[j]<=pivot:
                i+=1
                arr[i],arr[j]=arr[j],arr[i]
        arr[i+1],arr[high]=arr[high],arr[i+1]
        return i+1
    def quicksort(self,arr,low,high):
        if low<high:
            pi=self.partition(arr,low,high)
            self.quicksort(arr,low,pi-1)
            self.quicksort(arr,pi+1,high)
        return arr
arr = [10, 7, 8, 9, 1, 5]
low=0
high=len(arr)
print(high)
res=QuickSort()
print(res.quicksort(arr,low,high-1))
```

```python
#QuickSort for th pivot value low
class QuickSort:
    def partition(self,arr,low,high):
        pivot=arr[low]
        i=low
        for j in range(low+1,high+1):
            if arr[j]<=pivot:
                i+=1
                arr[i],arr[j]=arr[j],arr[i]
        arr[i],arr[low]=arr[i],arr[low]
        return i

    def quicksort(self,arr,low,high):
        if low<high:
            pi=self.partition(arr,low,high)
            self.quicksort(arr,low,pi-1)
            self.quicksort(arr,pi+1,high)
        return arr
arr = [10, 7, 8, 9, 1, 5]
low=0
high=len(arr)
print(high)
res=QuickSort()
print(res.quicksort(arr,low,high-1))
```

```python
class MergeSort:
    def mergeSort(self,arr):
        n=len(arr)
        if n>1:
            m=n//2
            n1=arr[:m]
            n2=arr[m:]
            l1=len(n1)
            l2=len(n2)
            self.mergeSort(n1)
            self.mergeSort(n2)
            i=j=k=0
            while i<l1 and j<l2:
                if n1[i]<n2[j]:
                    arr[k]=n1[i]
                    i+=1
                else:
                    arr[k]=n2[j]
                    j+=1
                k+=1
            while i<l1:
                arr[k]=n1[i]
                i+=1
                k+=1
            while j<l2:
                arr[k]=n2[j]
                j+=1
                k+=1
        return arr

arr = [10, 7, 8, 9, 1, 5]
res=MergeSort()
print(res.mergeSort(arr))










```

```python
class HeapSort:
    def heapify(self,arr,n,i):
        largest=i
        left=2*i+1
        right=2*i+2
        if left<n and arr[left]>arr[largest]:
            largest=left
        if right<n and arr[right]>arr[largest]:
            largest=right
        if largest!=i:
            arr[i],arr[largest]=arr[largest],arr[i]
            self.heapify(arr,n,largest)
    def sort(self,arr):
        n=len(arr)
        for i in range(n//2-1,-1,-1):
            self.heapify(arr,n,i)
        for i in range(n-1,0,-1):
            arr[0],arr[i]=arr[i],arr[0]
            self.heapify(arr,i,0)
        return arr
arr = [12, 11, 13, 5, 6, 7]
res=HeapSort()
print(res.sort(arr))
```

```python
class InsertionSort:
    def insertionSort(self,arr):
        for i in range(1,len(arr)):
            key=arr[i]
            j=i-1
            while j>=0 and arr[j]>key:
                arr[j+1]=arr[j]
                j-=1
            arr[j+1]=key
        return arr
arr=[12,11,13,5,6]
res=InsertionSort()
print(res.insertionSort(arr))
```

```python
class BubbleSort:
    def  sort(self,arr):
        n=len(arr)
        for i  in range (n):
            swapped=False
            for j in   range(n-1-i):
                if arr[j]>arr[j+1]:
                    arr[j],arr[j+1]=arr[j+1],arr[j]
                    swapped=True
            if swapped==False:
                break
        return arr
arr = [64, 34, 25, 12, 22, 11, 90]
res=BubbleSort()
print(res.sort(arr))
```
