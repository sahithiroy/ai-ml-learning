```python
'''In data structures and algorithms, a string is a contiguous sequence of characters (like letters, numbers, symbols, or whitespace) stored in memory. While often treated as a simple data type in high-level programming, strings are fundamentally specialized array data structures that come with unique memory behaviors, operations, and algorithmic patterns.1. Memory Representation & PropertiesImmutable vs. Mutable StringsHow strings behave in memory depends heavily on the programming language:Immutable Strings (Python, Java, JavaScript): Once created, a string cannot be modified in-place. Any string modification (like concatenation, slicing, or replacing characters) creates an entirely new string object in memory.Consequence: Repeated string concatenation in a loop (e.g., s += char) inside a loop takes $O(N^2)$ time because the entire string is copied on every iteration.Fix: Collect characters in a list/array and join them at the end ("".join(list)) in $O(N)$ time.Mutable Strings (C++, C, Rust String): Strings can be altered directly in memory without reallocating a new array, enabling $O(1)$ character overwrites.ASCII vs. UnicodeASCII: Uses 1 byte (8 bits) per character, encoding 128 standard characters (English alphabet, numbers, basic symbols).Unicode / UTF-8: Variable-length encoding (1 to 4 bytes per character) capable of representing characters from all global languages, emojis, and symbols.2. Core String Operations & Time ComplexitiesOperationDescriptionTime ComplexityNotes / Python ExampleAccess by IndexRead character at position i$O(1)$char = s[i]Length / SizeReturn total number of characters$O(1)$len(s) (stored as metadata)Search (Substrings)Search for a substring of length $M$ in string of length $N$$O(N \cdot M)$ naive / $O(N)$ optimals.find("abc") or in operatorConcatenationJoin string $A$ (length $N$) and $B$ (length $M$)$O(N + M)$Creates a new string in immutable languagesSlicingExtract substring from index $i$ to $j$$O(K)$ where $K = j - i$s[i:j] creates a new string copy3. Essential String Algorithm PatternsString problems in coding interviews almost always boil down to one of these 6 primary patterns:                              ┌───────────────────────────┐
                              │ STRING ALGORITHM PATTERNS │
                              └─────────────┬─────────────┘
                                            │
     ┌─────────────────┬────────────────────┼──────────────────┬─────────────────┬─────────────────┐
     ▼                 ▼                    ▼                  ▼                 ▼                 ▼
1. Two Pointers    2. Frequency Map    3. Sliding Window   4. Prefix/Suffix   5. String Tries   6. Pattern Match
(Palindrome/Rev)   (Anagrams/Counts)   (Substrings/K)     (Palindromic)      (AutoComplete)    (KMP/Rabin-Karp)
Pattern 1: Two Pointers (Inward or Outward)When to use: Palindrome validation, reversing strings, matching characters from both ends.How it works: One pointer starts at index 0 (left), another at index len(s) - 1 (right), and they move toward each other.Example Problems: Valid Palindrome, Reverse String.Pythondef isPalindrome(s: str) -> bool:
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True
Pattern 2: Frequency Map / Character ArrayWhen to use: Checking for Anagrams, counting character distributions, or checking if two strings share the exact same character counts.How it works: Instead of a full hash map, you can use a fixed-size array of size 26 (for lowercase English letters 'a' through 'z') or size 128/256 (for ASCII).Pythondef isAnagram(s: str, t: str) -> bool:
    if len(s) != len(t):
        return False
    
    counts = [0] * 26
    for char_s, char_t in zip(s, t):
        counts[ord(char_s) - ord('a')] += 1
        counts[ord(char_t) - ord('a')] -= 1
        
    return all(c == 0 for c in counts)
Pattern 3: Sliding Window (Dynamic Substrings)When to use: Finding the longest, shortest, or optimal substring satisfying a condition (e.g., maximum length substring with no repeating characters).How it works: Maintain two pointers (left and right) defining a "window". Expand right to explore, and shrink left when constraints are violated.Example Problems: Longest Substring Without Repeating Characters, Minimum Window Substring.Pythondef lengthOfLongestSubstring(s: str) -> int:
    char_map = {}
    left = max_len = 0
    
    for right, char in enumerate(s):
        if char in char_map and char_map[char] >= left:
            left = char_map[char] + 1  # Shrink window past duplicate
        char_map[char] = right
        max_len = max(max_len, right - left + 1)
        
    return max_len
Pattern 4: Advanced String Data Structures (Trie / Prefix Tree)When to use: Storing a dictionary of words for fast prefix lookups, autocomplete engines, or spell-checkers.Key Feature: Nodes store character edges; common prefixes share memory pathways.Lookup Time: $O(L)$ where $L$ is the length of the target word.Pattern 5: String Matching AlgorithmsWhen searching for a pattern $P$ inside a large text $T$:AlgorithmTime ComplexityBest Used ForNaive Search$O(N \cdot M)$Simple brute-force checksKnuth-Morris-Pratt (KMP)$O(N + M)$Pattern matching without backtracking using a Longest Prefix Suffix (LPS) tableRabin-Karp$O(N + M)$Substring search using Rolling Hash (great for multi-pattern search)Z-Algorithm$O(N + M)$Preprocessing prefixes for pattern lookup4. Cheat Sheet: Problem Keywords vs. PatternProblem Phrase / KeywordRecommended Pattern / Tool"Check if valid palindrome"Two Pointers (left & right)"Check if anagram", "Group anagrams"Frequency Array ([0] * 26) or Sorted key"Longest/shortest substring with..."Sliding Window (left & right expanding window)"Prefix matching", "Autocomplete", "Word Search"Trie (Prefix Tree)"Find pattern in large text"KMP Algorithm or Rabin-Karp Rolling Hash
'''
```
