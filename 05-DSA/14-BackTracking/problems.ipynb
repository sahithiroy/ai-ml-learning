```python
'''What is Backtracking?

Backtracking is a problem-solving technique where we:

Make a choice → explore it → if it doesn't work, undo the choice → try another choice.

Think about solving a maze.

6

You start at the entrance:

START
  ↓
Choose a path
  ↓
Is this path valid?
  ↓
YES → Continue
  ↓
NO → GO BACK
       ↓
    Try another path

That GO BACK is the key idea behind backtracking.

1. Why do we need backtracking?

Backtracking is useful when a problem says something like:

"Find all possible solutions"
"Generate all combinations"
"Generate all permutations"
"Find a valid arrangement"
"Try every possible choice"
"Find a path satisfying some conditions"
"Place items so that certain rules are satisfied"

In other words:

When there are many possible choices and you need to explore them systematically, backtracking is often useful.

2. A very simple example

Suppose you have:

[1, 2, 3]

You want to generate all subsets.

For each number, you have two choices:

Take it
Don't take it

For 1:

                []
              /    \
           Take 1  Skip 1

If we take 1:

                []
              /    \
            [1]     []
           /   \
       Take 2  Skip 2

Continue this process.

Eventually:

[]
[1]
[2]
[3]
[1,2]
[1,3]
[2,3]
[1,2,3]

We explored all choices.

3. Where does "backtracking" happen?

Suppose we have:

current = []

We decide:

current.append(1)

Now:

current = [1]

We explore all possibilities with 1.

When we're finished, we need to go back:

current.pop()

Now:

current = []

Then we can try another choice.

So:

Choose
  ↓
[1]
  ↓
Explore
  ↓
Finished
  ↓
UNDO
  ↓
[]
  ↓
Try another choice

That's backtracking.

4. The basic backtracking template

Almost every backtracking problem follows this general structure:

def backtrack(state):


    if solution_found:
        save_solution()
        return


    for choice in choices:


        if choice_is_valid:


            make_choice()


            backtrack(new_state)


            undo_choice()

The most important three operations are:

make choice
     ↓
recursive call
     ↓
undo choice

For example:

current.append(x)


backtrack(...)


current.pop()
5. When should I use backtracking?

When you see a problem involving choices + constraints + search, think about backtracking.

Example 1: Subsets

Question:

Find all subsets of [1,2,3].

At every element:

Take it
Don't take it

➡️ Backtracking.

Example 2: Permutations

Question:

Generate all arrangements of [1,2,3].

Choices:

Choose 1
Choose 2
Choose 3

Then continue choosing from the remaining elements.

➡️ Backtracking.

Example 3: N-Queens

Question:

Place N queens on an N×N chessboard so that no two queens attack each other.

Try a position:

Is it valid?
   |
   ├── YES → continue
   |
   └── NO → undo → try another position

➡️ Backtracking.

Example 4: Sudoku

Choose a number:

Try 5

Check:

Is 5 valid?

If yes:

Continue

If eventually you get stuck:

Remove 5
Try another number

➡️ Backtracking.

Example 5: Maze

At a cell:

Can I go UP?
Can I go DOWN?
Can I go LEFT?
Can I go RIGHT?

Try one direction.

If you reach a dead end:

Go back
Try another direction

➡️ Backtracking.

6. Backtracking vs normal recursion

This is important.

Normal recursion

Example:

def factorial(n):


    if n == 1:
        return 1


    return n * factorial(n-1)

There isn't really a choice.

It just does:

5
↓
4
↓
3
↓
2
↓
1
Backtracking

There are choices:

                 Start
               /       \
           Choice A   Choice B
            /   \       /   \
           ... ...    ...   ...

We explore one branch, then return and explore another.

7. Backtracking vs divide and conquer

These are often confused.

Divide and conquer

You divide the problem into independent subproblems.

Example:

                [1 2 3 4]
                /       \
             [1 2]     [3 4]

Merge Sort does this.

You solve both parts.

Backtracking

You are exploring choices.

Example:

                 Start
                /     \
             Take     Don't take
              /          \
            ...          ...

You don't necessarily need to explore every branch because invalid branches can be stopped early.

That stopping is called pruning.

8. What is pruning?

Pruning means:

Stop exploring a branch as soon as you know it cannot produce a valid solution.

Suppose we're solving N-Queens.

Place Queen
     ↓
Is position safe?
     ↓
    NO
     ↓
STOP THIS BRANCH

We don't waste time exploring the rest of that branch.

Example:

                 Start
               /       \
            Valid      Invalid ❌
             /           \
           ...          STOP

This can make backtracking much faster in practice.

9. How to solve a backtracking problem

Use these 5 steps.

Step 1: Identify the choices

Ask:

"At this point, what choices can I make?"

For subsets:

Take
Don't take

For N-Queens:

Column 0
Column 1
Column 2
...
Column N-1

For Sudoku:

1,2,3,4,5,6,7,8,9
Step 2: Identify the constraint

Ask:

"Which choices are valid?"

For N-Queens:

No same column
No same diagonal
No same row

For Sudoku:

No duplicate in row
No duplicate in column
No duplicate in 3×3 box
Step 3: Make the choice

For example:

current.append(x)

or:

board[row][col] = 'Q'
Step 4: Recursively explore
backtrack(...)

Now we solve the smaller remaining problem.

Step 5: Undo the choice

This is the most important step:

current.pop()

or:

board[row][col] = '.'

Then try another choice.

10. Let's solve a simple problem
Problem

Given:

[1, 2]

Generate all subsets.

Start:

current = []

For 1 we have two choices.

Choice 1: Don't take 1
[]

Then process 2.

Don't take:

[]

Take:

[2]

So we get:

[]
[2]

Now we backtrack to the point before processing 1.

Choice 2: Take 1
[1]

Then process 2.

Don't take:

[1]

Take:

[1,2]

So we get:

[1]
[1,2]

Final:

[]
[1]
[2]
[1,2]
11. Code
def subsets(arr, index, current):


    # Base case
    if index == len(arr):
        print(current.copy())
        return


    # Choice 1: Don't take arr[index]
    subsets(arr, index + 1, current)


    # Choice 2: Take arr[index]
    current.append(arr[index])


    subsets(arr, index + 1, current)


    # BACKTRACK
    current.pop()

The important section is:

current.append(arr[index])


subsets(arr, index + 1, current)


current.pop()

Think:

        MAKE CHOICE
             ↓
      RECURSIVE CALL
             ↓
        UNDO CHOICE
12. How to recognize backtracking in coding questions

Look for phrases such as:

🚨 Strong backtracking signals

"Generate all..."

"Find all possible..."

"Return all combinations..."

"Return all permutations..."

"Find all paths..."

"Place N items..."

"Try every possibility..."

"Find a valid arrangement..."

"Choose numbers such that..."

"No two elements can..."

"If the choice is invalid, try another..."

These are strong signals that backtracking may be appropriate.

13. Backtracking complexity

Once you understand the algorithm, then calculate complexity.

A useful way to think about it is:

Time=number of states/nodes explored×work at each state
	​

Subsets

Two choices at every level:

2
N

So basic recursion:

O(2
N
)

If copying each subset takes O(N):

O(N2
N
)
Permutations

Choices:

N
N-1
N-2
...
1

So:

N!

If copying each permutation takes O(N):

O(N⋅N!)
Space

Usually:

Recursion depth + current state
	​


For subsets:

O(N)

For permutations:

O(N)

If you store all answers, output space must also be counted.

14. The easiest way to remember backtracking

Remember this sentence:

Backtracking is DFS over a decision tree.

And the basic pattern:

                 CHOICES
                    ↓
                 CHOOSE
                    ↓
                EXPLORE
                    ↓
                 UNDO
                    ↓
             NEXT CHOICE

Or simply:

Choose → Explore → Unchoose
Your learning path

Since you're learning recursion and complexity, I would recommend practicing backtracking in this order:

1. Subsets
      ↓
2. Permutations
      ↓
3. Combinations
      ↓
4. Combination Sum
      ↓
5. Generate Parentheses
      ↓
6. Rat in a Maze
      ↓
7. Word Search
      ↓
8. N-Queens
      ↓
9. Sudoku

Once you understand Subsets, the rest becomes much easier because the same choose → explore → undo pattern keeps appearing.'''
```

```python
'''78. Subsets
Solved
Medium
Topics
premium lock icon
Companies
Given an integer array nums of unique elements, return all possible subsets (the power set).

The solution set must not contain duplicate subsets. Return the solution in any order.

 

Example 1:

Input: nums = [1,2,3]
Output: [[],[1],[2],[1,2],[3],[1,3],[2,3],[1,2,3]]
Example 2:

Input: nums = [0]
Output: [[],[0]]
 

Constraints:

1 <= nums.length <= 10
-10 <= nums[i] <= 10
All the numbers of nums are unique.'''
class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        res=[]
        curr=[]
        def backtrack(index):
            if index==len(nums):
                res.append(curr[:])
                return
            curr.append(nums[index])
            backtrack(index+1)
            curr.pop()
            backtrack(index+1)
            
        backtrack(0)
        return res
        
```

```python
'''90. Subsets II
Solved
Medium
Topics
premium lock icon
Companies
Given an integer array nums that may contain duplicates, return all possible subsets (the power set).

The solution set must not contain duplicate subsets. Return the solution in any order.

 

Example 1:

Input: nums = [1,2,2]
Output: [[],[1],[1,2],[1,2,2],[2],[2,2]]
Example 2:

Input: nums = [0]
Output: [[],[0]]
 

Constraints:

1 <= nums.length <= 10
-10 <= nums[i] <= 10'''
class Solution:
    def subsetsWithDup(self, nums: List[int]) -> List[List[int]]:
        res=[]
        curr=[]
        nums.sort()
        def backtrack(i):
            if i==len(nums):
                
                res.append(curr[:])
                return
            curr.append(nums[i])
            backtrack(i+1)
            curr.pop()
            while i+1<len(nums) and nums[i+1]==nums[i]:
                i+=1
            backtrack(i+1)
        backtrack(0)
        return res       
```

```python
'''Generate all binary strings
Solved
Difficulty: MediumAccuracy: 63.9%Submissions: 51K+Points: 4Average Time: 20m
Given an integer n. You need to generate all the binary strings of n characters representing bits.

Note: Return the strings in  ascending order.

Examples:

Input: n = 2
Output: [00, 01, 10, 11]
Explanation: As each position can be either 0 or 1, the total possible combinations are 4.
Input: n = 3
Output: [000, 001, 010, 011, 100, 101, 110, 111]
Explanation: As each position can be either 0 or 1, the total possible combinations are 8.'''
class Solution:
    def binstr(self, n):
        # code here
        res=[]
        def dfs(curr):
            if len(curr)==n:
                res.append(curr)
                return
            dfs(curr+"0")
            dfs(curr+ "1")
        dfs("")
        return res
```

```python
'''77. Combinations
Solved
Medium
Topics
premium lock icon
Companies
Given two integers n and k, return all possible combinations of k numbers chosen from the range [1, n].

You may return the answer in any order.

 

Example 1:

Input: n = 4, k = 2
Output: [[1,2],[1,3],[1,4],[2,3],[2,4],[3,4]]
Explanation: There are 4 choose 2 = 6 total combinations.
Note that combinations are unordered, i.e., [1,2] and [2,1] are considered to be the same combination.
Example 2:

Input: n = 1, k = 1
Output: [[1]]
Explanation: There is 1 choose 1 = 1 total combination.'''
class Solution:
    def combine(self, n: int, k: int) -> List[List[int]]:
        res=[]
        curr=[]
        def dfs(i):
            if len(curr)==k:
                res.append(curr[:])
                return
            for num in range(i,n+1):
                curr.append(num)
                dfs(num+1)
                curr.pop()
        dfs(1)
        return res        
```
