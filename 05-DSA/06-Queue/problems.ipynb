```python
'''A Queue is a linear data structure that follows the FIFO (First In, First Out) principle.Think of it like a real-life queue at a movie ticket counter or a bus stop:The first person to stand in line is the first person to get a ticket and leave.New people join the line at the back (rear).People leave the line from the front.5 Key Queue OperationsOperationStandard TermWhat It DoesPython (collections.deque)EnqueueAddInserts an element at the rear (back)q.append(item)DequeueRemoveRemoves and returns the element at the frontq.popleft()Front / PeekView FrontLooks at the front element without removing itq[0]isEmptyCheck EmptyChecks if the queue has no elementsnot q or len(q) == 0SizeSizeReturns total number of elements in the queuelen(q)All primary operations (Enqueue and Dequeue) run in $O(1)$ constant time.How to Implement a Queue in PythonIn Python, do NOT use a regular list for a queue. Removing an item from the front of a list (list.pop(0)) takes $O(n)$ time because all remaining elements have to shift left in memory.Instead, Python provides collections.deque (Double-Ended Queue), which performs $O(1)$ operations at both ends.Recommended Implementation (collections.deque)Pythonfrom collections import deque

# 1. Initialize an empty queue
queue = deque()

# 2. ENQUEUE: Add elements to the rear
queue.append("Alice")
queue.append("Bob")
queue.append("Charlie")
print("Queue after enqueues:", queue)  # deque(['Alice', 'Bob', 'Charlie'])

# 3. PEEK: Look at the front element
print("Front element:", queue[0])       # Alice

# 4. DEQUEUE: Remove element from the front
first_out = queue.popleft()
print("Dequeued element:", first_out)  # Alice
print("Queue after dequeue:", queue)   # deque(['Bob', 'Charlie'])

# 5. IS_EMPTY: Check if queue is empty
print("Is queue empty?", len(queue) == 0) # False
Common Types of QueuesStandard Queue (FIFO): Items enter at the rear and exit from the front.Circular Queue: The last position is connected back to the first position to reuse empty spaces created by dequeuing.Double-Ended Queue (Deque): Elements can be added or removed from both the front and rear.Priority Queue: Elements are dequeued based on priority rather than arrival order (e.g., emergency room patients). Implemented in Python using heapq.When do you use a Queue?Breadth-First Search (BFS): Traversing graphs or trees level-by-level.Task Scheduling: CPU task scheduling, printer print jobs.Buffering: Streaming audio/video playback (buffering data packets in arrival order).'''
```

```python
'''Given an array arr[]  and a positive integer k, find the first negative integer for each and every window(contiguous subarray) of size k.

Note: If a window does not contain a negative integer, then return 0 for that window.

Examples:

Input: arr[] = [-8, 2, 3, -6, 10] , k = 2
Output: [-8, 0, -6, -6]
Explanation:
Window [-8, 2] First negative integer is -8.
Window [2, 3] No negative integers, output is 0.
Window [3, -6] First negative integer is -6.
Window [-6, 10] First negative integer is -6.
Input: arr[] = [12, -1, -7, 8, -15, 30, 16, 28] , k = 3
Output: [-1, -1, -7, -15, -15, 0] 
Explanation:
Window [12, -1, -7] First negative integer is -1.
Window [-1, -7, 8] First negative integer is -1.
Window [-7, 8, -15] First negative integer is -7.
Window [8, -15, 30] First negative integer is -15.
Window [-15, 30, 16] First negative integer is -15.
Window [30, 16, 28] No negative integers, output is 0.
Input: arr[] = [12, 1, 3, 5] , k = 3
Output: [0, 0] 
Explanation:
Window [12, 1, 3] No negative integers, output is 0.
Window [1, 3, 5] No negative integers, output is 0.'''
from collections import deque
class Solution:
    def firstNegInt(self, arr, k): 
         # code here 
        q=deque()
        res=[]
        for i,num in enumerate(arr):
            if num<0:
                q.append(i)
            if q and q[0]<=i-k:
                q.popleft()
                
            if i>=k-1:
                if len(q)>=1:
                    res.append(arr[q[0]])
                if len(q)==0:
                    res.append(0)
        return res
                    
                
```

```python
'''Given an array arr[] of positive integers and an integer k. Find the maximum value for each contiguous subarray of size k.

Return an array of maximum values corresponding to each contiguous subarray.

Examples:

Input: arr[] = [1, 2, 3, 1, 4, 5, 2, 3, 6], k = 3
Output: [3, 3, 4, 5, 5, 5, 6]
Explanation: 
1st contiguous subarray [1, 2, 3], max = 3
2nd contiguous subarray [2, 3, 1], max = 3
3rd contiguous subarray [3, 1, 4], max = 4
4th contiguous subarray [1, 4, 5], max = 5
5th contiguous subarray [4, 5, 2], max = 5
6th contiguous subarray [5, 2, 3], max = 5
7th contiguous subarray [2, 3, 6], max = 6
Input: arr[] = [5, 1, 3, 4, 2], k = 1
Output: [5, 1, 3, 4, 2]
Explanation: When k = 1, each element in the array is its own subarray, so the output is simply the same array'''
from collections import deque
class Solution:
    def maxOfSubarrays(self, arr, k):
        # code here
        q=deque()
        res=[]
        for i,num in enumerate(arr):
            while q and arr[q[-1]]<arr[i]:
                q.pop()
            q.append(i)
            if q[0]<=i-k:
                q.popleft()
            if i>=k-1:
                res.append(arr[q[0]])
        return res
                
        
```
