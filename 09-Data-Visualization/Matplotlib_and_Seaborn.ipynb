```python
#Multiple Points
xpoints = np.array([1, 2, 6, 8])
ypoints = np.array([3, 8, 1, 10])

plt.plot(xpoints, ypoints)
plt.show()
```

```python
#Mention the marker size
plt.plot(xpoints,ypoints,marker='o',ms=20)
plt.show()
```

```python
#mention the marker Color mfc=markerfacecolor
plt.plot(xpoints,ypoints,marker='o',ms=20,mfc='r')
plt.show()
```

```python
#Mention the marker color mec marker edge color
plt.plot(xpoints,ypoints,marker='o',ms=20,mfc='r',mec='b')
plt.show()
```

```python
#MentionLine style -->dotted,dashed
plt.plot(xpoints,ypoints,marker='o',ms=20,mfc='r',mec='b',linestyle='dotted')
plt.show()
```

```python
#Mention the labels
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.plot(x, y)
plt.title("Sports Watch Data")
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.show()
```

```python
#Font Properties
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

font1 = {'family':'serif','color':'blue','size':20}
font2 = {'family':'serif','color':'darkred','size':15}

plt.title("Sports Watch Data", fontdict = font1)
plt.xlabel("Average Pulse", fontdict = font2)
plt.ylabel("Calorie Burnage", fontdict = font2)

plt.plot(x, y)
plt.show()
```

```python
#Mention the position of the title
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.title("Sports Watch Data", loc = 'left')
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.plot(x, y)
plt.show()
```

```python
'''A histogram is a graph showing frequency distributions.

It is a graph showing the number of observations within each given interval.

Example: Say you ask for the height of 250 people, you might end up with a histogram like this:
'''
x = np.random.normal(170, 10, 250)

plt.hist(x)
plt.show()
```

```python
#Pie Charts
y = np.array([35, 25, 25, 15])

plt.pie(y)
plt.show()
```

```python
y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.pie(y, labels = mylabels)
plt.show()
```

```python
#Mention Explode,legend and sahdow
y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
myexplode = [0.2, 0, 0, 0]

plt.pie(y, labels = mylabels, explode = myexplode,shadow=True)
plt.legend(title = "Four Fruits:")
plt.show()
```

```python
import seaborn as sns
import pandas as pd

# Sample dataframe
df = pd.DataFrame({
    "x": [1, 2, 3, 4, 5],
    "y": [2, 4, 6, 8, 10]
})

sns.set_style("whitegrid")
sns.lineplot(data=df, x="x", y="y", marker="o")
```

```python
''' set Styles:
darkgrid
whitegrid
dark
white
ticks
'''
#Multiple PLots
# loading dataset
data = sns.load_dataset("iris")

def graph():
    sns.lineplot(x="sepal_length", y="sepal_width", data=data)

# Creating a new figure with width = 5 inches
# and height = 4 inches
fig = plt.figure(figsize =(5, 4))

# Creating first axes for the figure
ax1 = fig.add_axes([0.1, 0.1, 0.8, 0.8])

# plotting the graph
graph()

# Creating second axes for the figure
ax2 = fig.add_axes([0.5, 0.5, 0.3, 0.3])

# plotting the graph
graph()

plt.show()
```
